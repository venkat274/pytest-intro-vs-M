```python
df = df.drop("date")
```

```python
df.show()
```

```python
df = df.withColumn("from_date",lit("12/1/10 08:26"))
#df = df.withColumn('from_date',to_timestamp("from_date", 'yy/MM/dd HH:mm'))
df = df.withColumn('date',df["InvoiceDate"])


df2 = df.withColumn('from_date',to_timestamp(col('from_date'))).withColumn('recency',col('InvoiceDate').cast("long") - col('from_date').cast("long"))
df2.show(5,0)
```

```python
df2 = df2.join(df2.groupBy('CustomerID').agg(max('recency').alias('recency')),on='recency',how='leftsemi')
df2.show()
```

```python
df2.printSchema()
```

```python
df_freq = df.groupBy('CustomerID').agg(count('InvoiceDate').alias('frequency'))
df_freq.show(5,0)
```

```python
df3 = df.join(df_freq,on='CustomerID',how='inner')
df3.count()
```

```python
df3.printSchema()
```

```python
m_val = df3.withColumn('Total amount',col("Quantity") * col("UnitPrice"))
```

```python
m_val = m_val.groupBy("CustomerID").agg(sum("UnitPrice")).alias("monetory value")
```

```python
finaldf = m_val.join(df3,on='CustomerID',how='inner')
```

```python
finaldf.show(5,0)
```
